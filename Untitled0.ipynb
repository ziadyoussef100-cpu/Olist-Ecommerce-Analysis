```python
orders = pd.read_csv('/content/olist_orders_dataset.csv')
```

```python
print("Orders:", orders.shape)

print("\nMissing Values:")
print(orders.isnull().sum())

print("\nDuplicates:", orders.duplicated().sum())
```

```python
orders['order_status'].value_counts()
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])

orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
```

```python
orders['delivery_days'] = (
    orders['order_delivered_customer_date'] -
    orders['order_purchase_timestamp']
).dt.days
```

```python
orders['delivery_days'].mean()
```

```python
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])

delivered_orders = orders[
    orders['order_delivered_customer_date'].notna()
]

delayed_orders = delivered_orders[
    delivered_orders['order_delivered_customer_date'] >
    delivered_orders['order_estimated_delivery_date']
]

len(delayed_orders) / len(delivered_orders) * 100
```

```python
import pandas as pd

customers = pd.read_csv('/content/drive/MyDrive/olist_customers_dataset.csv')
orders = pd.read_csv('/content/drive/MyDrive/olist_orders_dataset.csv')
order_items = pd.read_csv('/content/drive/MyDrive/olist_order_items_dataset.csv')
payments = pd.read_csv('/content/drive/MyDrive/olist_order_payments_dataset.csv')
products = pd.read_csv('/content/drive/MyDrive/olist_products_dataset.csv')
sellers = pd.read_csv('/content/drive/MyDrive/olist_sellers_dataset.csv')
geolocation = pd.read_csv('/content/drive/MyDrive/olist_geolocation_dataset.csv')
category_translation = pd.read_csv('/content/drive/MyDrive/product_category_name_translation.csv')
```

```python
product_sales = order_items.merge(products, on='product_id')

product_sales.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)
```

```python
sales_by_state = order_items.merge(
    orders[['order_id', 'customer_id']], on='order_id'
)

sales_by_state = sales_by_state.merge(
    customers[['customer_id', 'customer_state']], on='customer_id'
)

sales_by_state.groupby('customer_state')['price'].sum().sort_values(ascending=False).head(10)
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

monthly_sales = order_items.merge(
    orders[['order_id', 'order_purchase_timestamp']],
    on='order_id'
)

monthly_sales['month'] = monthly_sales['order_purchase_timestamp'].dt.to_period('M')

monthly_sales.groupby('month')['price'].sum()
```
